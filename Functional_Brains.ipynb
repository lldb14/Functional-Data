```python
import os
import requests
import hashlib
from tqdm import tqdm
from google.colab import drive

# --- CONFIGURACIÓN ---
RECORD_ID = "12677223"
# Directorio donde guardar los archivos (cámbialo si quieres)
SAVE_DIR = "/content/drive/MyDrive/MedIAnomaly_Data"

# Lista de archivos a descargar con sus MD5 (según la imagen)
# Si algún MD5 no es correcto, se mostrará una advertencia.
FILES_TO_DOWNLOAD = {
    "BrainTumor.tar.gz": "ab6482f646262fc4669b0135d80a0482",
    "BraTS2021.tar.gz": "237f2aee77e099f7483808ddec49a57c",
    "Camelyon16.tar.gz": "c26c026d1c315deb52720d4c195e874f",
    "LAG.tar.gz": "0b4034ae8d25757723a512adb218f86a",
    "RSNA.tar.gz": "5db85ce4b2daad8f1760eb23c050a7d",
    "VinCXR.tar.gz": "45cf848a4eba925062531c5f2d344ddc"
}
# --- FIN CONFIGURACIÓN ---

def get_download_link(record_id, filename):
    """Construye el enlace directo de descarga para un archivo específico."""
    return f"https://zenodo.org/records/{record_id}/files/{filename}?download=1"

def download_file(url, dest_path):
    """Descarga un archivo con barra de progreso."""
    try:
        response = requests.get(url, stream=True)
        response.raise_for_status()
        total_size = int(response.headers.get('content-length', 0))
        progress = tqdm(total=total_size, unit='B', unit_scale=True, desc=os.path.basename(dest_path))
        with open(dest_path, 'wb') as f:
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)
                progress.update(len(chunk))
        progress.close()
        return True
    except Exception as e:
        print(f"❌ Error en descarga: {e}")
        return False

def verify_md5(file_path, expected_md5):
    """Calcula el MD5 del archivo y lo compara con el esperado."""
    hash_md5 = hashlib.md5()
    with open(file_path, "rb") as f:
        for chunk in iter(lambda: f.read(4096), b""):
            hash_md5.update(chunk)
    actual_md5 = hash_md5.hexdigest()
    if actual_md5 == expected_md5:
        print(f"✅ MD5 correcto: {actual_md5}")
        return True
    else:
        print(f"⚠️ MD5 incorrecto. Esperado: {expected_md5}, obtenido: {actual_md5}")
        return False

# --- EJECUCIÓN ---
print("🔗 Montando Google Drive...")
drive.mount('/content/drive')
os.makedirs(SAVE_DIR, exist_ok=True)
print(f"📁 Los archivos se guardarán en: {SAVE_DIR}\n")

for filename, expected_md5 in FILES_TO_DOWNLOAD.items():
    save_path = os.path.join(SAVE_DIR, filename)
    if os.path.exists(save_path):
        print(f"⏭️ {filename} ya existe. Verificando MD5...")
        verify_md5(save_path, expected_md5)
        continue

    print(f"📥 Descargando {filename}...")
    url = get_download_link(RECORD_ID, filename)
    if download_file(url, save_path):
        print("🔍 Verificando integridad...")
        if verify_md5(save_path, expected_md5):
            print(f"✅ {filename} descargado y verificado.\n")
        else:
            print(f"❌ {filename} se descargó pero el MD5 no coincide. Revisa la descarga.\n")
    else:
        print(f"❌ Falló la descarga de {filename}\n")

print("🎉 Proceso completado.")
```

```python
import os
import tarfile

# Ruta donde guardaste los archivos .tar.gz (cambia si usaste otra)
SAVE_DIR = "/content/drive/MyDrive/MedIAnomaly_Data"
# Ruta donde se extraerá el contenido
EXTRACT_DIR = "/content/drive/MyDrive/MedIAnomaly_Data_extracted"

os.makedirs(EXTRACT_DIR, exist_ok=True)

# Lista de archivos que descargaste
tar_files = [
    "BrainTumor.tar.gz",
    "BraTS2021.tar.gz",
    "Camelyon16.tar.gz",
    "LAG.tar.gz",
    "RSNA.tar.gz",
    "VinCXR.tar.gz"
]

for tar_file in tar_files:
    tar_path = os.path.join(SAVE_DIR, tar_file)
    if os.path.exists(tar_path):
        print(f"📦 Extrayendo {tar_file} ...")
        with tarfile.open(tar_path, "r:gz") as tar:
            tar.extractall(path=EXTRACT_DIR)
        print(f"   ✅ Extraído en {EXTRACT_DIR}")
    else:
        print(f"⚠️ No se encontró {tar_file}")
```
